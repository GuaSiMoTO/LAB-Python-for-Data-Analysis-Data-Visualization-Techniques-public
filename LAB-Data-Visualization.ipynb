### Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Load the dataset

```python
# uso el encoding por buena practiva
df = pd.read_csv('retail_sales_data.csv', encoding ='UTF-8')

# paso los nombres de las columnas a minuscula
df.columns = df.columns.str.lower()
```

```python
### Visualize Monthly Sales
```

```python
#df['Date']= pd.to_datetime(df['Date'],format ='%Y-%m-%d')

#df['month'] = df['Date'].dt.month


df_total_sales = df.groupby('month')['sales'].sum().reset_index(name = 'total sales')

fig, ax = plt.subplots(1, 1)

ax.bar(df_total_sales['month'],df_total_sales['total sales'])

ax.set_title('Total Sales x Months')
ax.set_xlabel('months')
ax.set_ylabel('Total sales')

plt.show()
```

```python
plt.plot(df_total_sales['month'],df_total_sales['total sales'], marker = 'o', color = 'red')

plt.title('Seasonal Sales Trends')
plt.xlabel('MONTHS')
plt.ylabel('SALES')

plt.show()
```

### Analyze Sales Trends

### Explore Sales and Discounts Relationship

```python
fig, ax = plt.subplots(1, 1)

ax.scatter(df['discount'],df['sales'])

ax.set_xlabel('Discounts')
ax.set_ylabel('Total Sales')
ax.set_title('Sales and Discounts Relationship')

ax.annotate('YEAAAH!!',  # Texto a mostrar
            xy=(19, 1140),      # Punto a anotar (coordenadas del dato)
            xytext=(40, 1000), # Posición del texto
            arrowprops=dict(facecolor='red', arrowstyle ='->'))



plt.show()
```

## Iteration 2: Plotting with Seaborn

### Correlation Heatmap

```python
# o esta manera para pillar todas las columnas de tipo float o int
# corr = df.select_dtypes(include= [float, int]).corr()

corr = df[['sales', 'discount']].corr() # aqui solo pillo las dos columnas 'sales' y 'discount'

sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1)

plt.show()
```

### Product Category Analysis

### Sales Distribution by Category

```python
# estilo whitegrid
# sns.set(style="whitegrid")

# estilo darkgrid
sns.set(style = "darkgrid")

# Crear FacetGrid para histogramas por categoría
g = sns.FacetGrid(df, col="product category", col_wrap=2, height=5, sharex=False, sharey=False)
g.map(sns.histplot, "sales", kde=True, bins=15, color='silver')

# Añadir títulos y ajustes
g.set_titles("{col_name}")  # Usar nombres de categoría como títulos de cada grafica
g.set_axis_labels("Ventas", "Frecuencia")
g.fig.suptitle('Distribución de Ventas Producto')

plt.tight_layout()
plt.show()
```

### Regional Sales Analysis

```python
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))

# Agrupar datos por región y con sort ordenamos ascendente o descendiente añadiendo (ascending=False)
sales_region = df.groupby('region')['sales'].sum().reset_index().sort_values('sales')

# Crear gráfico de barras
ax = sns.barplot(data=sales_region, x='region', y='sales', palette="Blues")

# Personalización final
plt.title('Ventas Totales por Región', pad = 10)
plt.xlabel('Región')
plt.ylabel('Ventas Totales')
plt.tight_layout()
plt.show()
```
